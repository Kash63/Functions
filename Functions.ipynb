1. What is the difference between a function and a method in Python?
> n Python, the key distinction between a function and a method lies in their association. A function is a standalone block of code defined outside of any class, designed to perform a specific task and can be called directly by its name. On the other hand, a method is a function that is associated with an object of a class. It is defined within a class and is called on an object using dot notation. The first parameter of a method is conventionally self, which refers to the instance of the class on which the method is being called, allowing the method to access and modify the object's attributes. Essentially, methods are functions that operate within the context of a specific object.

5. What are iterators in Python and how do they differ from iterables?
> An iterator, is an object that enables the traversal of an iterable. It remembers its state (i.e., the current position) and has a __next__() method that returns the next item in the sequence. When there are no more items, the __next__() method raises a StopIteration exception. Iterators also implement the __iter__() method, which simply returns the iterator object itself, making them iterable as well.

> The difference is that an iterable is something you can get an iterator from, while an iterator is the object that allows you to actually iterate over the iterable. You can think of an iterable as a container, and an iterator as a cursor that moves through the container, providing access to each item.

7. What are the advantages of using generators over regular functions?
> The primary advantages of using generators over regular functions in Python stem from their lazy evaluation and state retention.

1  Regular functions compute all their results and store them in memory before returning, which can be memory-intensive, especially for large datasets or infinite sequences. In contrast, generators produce values on demand using the yield keyword, generating one item at a time only when requested.

2  This lazy evaluation significantly reduces memory usage, making generators ideal for processing large files or streams of data without loading everything into memory simultaneously.

3  Furthermore, generators maintain their internal state between calls.

 4  When next() is called on a generator, it resumes execution from where it left off, preserving the values of local variables.

  5  This simplifies the creation of iterators for complex sequences, as you don't need to manually manage the state within a class. This combination of memory efficiency and simplified state management makes generators a powerful tool for writing clean and scalable code in Python.

9.  Explain the purpose and usage of the `map()` function in Python.
> The map() function in Python serves the purpose of applying a given function to each item of an iterable and returning an iterator that yields the results. Its syntax is map, where function is the function to be applied and iterable is the sequence of items to process. The map() function iterates through each element in the iterable, calls the provided function with that element as an argument, and yields the return value of the function. This provides a concise way to perform the same operation on multiple items without writing explicit loops. For instance, to square every number in a list, you could use map(lambda x: x**2, [1, 2, 3]), which would return an iterator yielding 1, 4, and 9. You often need to explicitly convert the resulting iterator to a list or another sequence type to view the results directly. The map() function is particularly useful for applying transformations to collections of data in a functional programming style.

10.  What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?
>  The map(), filter(), and reduce() functions in Python are all higher-order functions that operate on iterables, but they serve distinct purposes. map() applies a given function to each item in an iterable and returns an iterator of the results, transforming each element individually. filter() also applies a given function to each item in an iterable, but it returns an iterator containing only the items for which the function returns True, effectively selecting elements based on a condition. reduce(), on the other hand, applies a function cumulatively to the items of an iterable, from left to right, so as to reduce the iterable to a single value. It 1  takes a function of two arguments and successively applies it to the items in the iterable, along with a running "result." For example, to find the product of all numbers in a list, you could use reduce(lambda x, y: x * y, [1, 2, 3, 4]). While map() and filter() produce iterators of the same length (or shorter in the case of filter()) as the input iterable, reduce() aims to condense the iterable into a single output value.

```python
'''4. Write a Python function that checks if a given number is prime or not from 1 to 200.'''
def prime(number):
    if not 1 <= number <= 200:
        return False
    if number <= 3:
        return number > 1
    if number % 2 == 0 or number % 3 == 0:
        return False
    for i in range(5, int(number**0.5) + 1, 6):
        if number % i == 0 or number % (i + 2) == 0:
            return False
    return True
```

```python
'''5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of
terms.'''
class FibonacciIterator:
    def __init__(self, num_terms):
        if not isinstance(num_terms, int) or num_terms <= 0:
            raise ValueError("num_terms must be a positive integer")
        self.num_terms = num_terms
        self.n1 = 0
        self.n2 = 1
        self.counter = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.counter < self.num_terms:
            result = self.n1
            nth = self.n1 + self.n2
            self.n1 = self.n2
            self.n2 = nth
            self.counter += 1
            return result
        else:
            raise StopIteration
```

```python
#7.Implement a generator function that reads a file line by line and yields each line as a string.
def read_file_lines(filename):
    try:
        with open(filename, 'r') as file:
            for line in file:
                yield line.strip()
    except FileNotFoundError:
        yield f"File not found: {filename}"
```

```python
#8. Use a lambda function in Python to sort a list of tuples based on the second element of each tuple.
def sort_tuples_by_second_element(list_of_tuples):
    return sorted(list_of_tuples, key=lambda x: x[1])
```

```python
#9. Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.
def celsius_to_fahrenheit(celsius_list):
    return map(lambda c: (c * 9/5) + 32, celsius_list)
```

```python
# 10.  Create a Python program that uses `filter()` to remove all the vowels from a given string.
def remove_vowels(text_string):
    vowels_to_remove = "AEIOUaeiou"
    return "".join(filter(lambda character: character not in vowels_to_remove, text_string))
```
